```python
from datetime import datetime

saldo = 0.0
extrato = ""

def main():
    while True:
        print("""
        \n M E N U
        1. Depositar
        2. Saque
        3. Ver saldo
        4. Ver extrato
        5. sair
        """)

        try:
            opc = int(input("Digite sua opção: "))
            if (opc < 1) or (opc > 5):
                print("\n Digite um numero válido")


            elif opc == 1:
                try:
                    valor = float(input("Digite o valor do depósito: R$"))
                    if valor <= 0:
                        print("O depósito não pode ser menor ou igual a 0")
                    else:
                        deposito(valor)
                except ValueError:
                    print("ERRO: Digite apenas números!")


            elif opc == 2:
                try:
                    retirada = float(input("Quanto vai retirar: R$ "))

                    if retirada <= 0:
                        print("O saque deve ser maior que zero!")
                    elif retirada > saldo:
                        print("O valor do saque não pode ser maior que o saldo!")
                    else:
                        saque(retirada)
                except ValueError:
                    print("ERRO: Digite apenas números!")


            elif opc == 3:
                vefsaldo()


            elif opc == 4:
                vefextrato()


            elif opc == 5:
                exit()
        except ValueError:
            print("ERRO: Só pode números")



def deposito(valor):
    global saldo
    global extrato

    agora = datetime.now().strftime("%d/%m/%Y %H:%M:%S")

    saldo = saldo + valor
    extrato = extrato + f"- Depósito efetuado no valor de {valor}\n" + agora

def saque(retirada):
    global saldo
    global extrato

    agora = datetime.now().strftime("%d/%m/%Y %H:%M:%S")

    saldo = saldo - retirada
    extrato = extrato + f"- Saque efetuado no valor de {retirada}" + agora

def vefsaldo():
    print("Saldo atual: R$", saldo)

def vefextrato():
    print("-- EXTRATO DA CONTA --")
    print(extrato)

if __name__ == '__main__':
    main()
```
